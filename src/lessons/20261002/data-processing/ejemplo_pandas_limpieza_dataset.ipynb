# Ejemplo de carga y limpieza de un dataset con Pandas

En este notebook vamos a trabajar con un archivo CSV que contiene datos ficticios de clientes y compras.

El objetivo es aprender a:

1. Cargar un archivo CSV con `pandas`.
2. Explorar rápidamente un dataset.
3. Detectar valores nulos y duplicados.
4. Limpiar textos inconsistentes.
5. Convertir columnas al tipo de dato correcto.
6. Detectar valores imposibles o anómalos.
7. Eliminar o corregir registros problemáticos.
8. Guardar el dataset limpio en un nuevo CSV.

> El archivo utilizado es **`ventas_clientes_sucio.csv`**.

## 1. Importar Pandas

`pandas` es una de las librerías más utilizadas en Python para trabajar con datos tabulares.

Normalmente se importa utilizando el alias `pd`.

````python
import pandas as pd
````

## 2. Cargar el archivo CSV

La función `pd.read_csv()` permite leer un archivo CSV y convertirlo en un **DataFrame**.

Un DataFrame es una estructura similar a una tabla de Excel: tiene filas y columnas.

````python
df = pd.read_csv("ventas_clientes_sucio.csv")

df
````

## 3. Primer vistazo al dataset

Antes de limpiar nada, conviene entender qué tenemos.

Algunos comandos básicos:

- `head()` → muestra las primeras filas.
- `shape` → indica el número de filas y columnas.
- `columns` → muestra los nombres de las columnas.
- `info()` → muestra tipos de datos y valores no nulos.
- `describe()` → resume las columnas numéricas.

````python
df.head()
````

````python
print("Dimensiones:", df.shape)
print("\nColumnas:")
print(df.columns.tolist())
````

````python
df.info()
````

### ¿Qué problemas podemos observar?

Este dataset contiene errores frecuentes en datos reales:

- espacios al principio o al final de algunos textos;
- ciudades escritas con distintas combinaciones de mayúsculas y minúsculas;
- valores nulos;
- filas duplicadas;
- edades almacenadas como texto;
- edades imposibles;
- importes escritos con coma decimal;
- importes negativos;
- texto dentro de una columna numérica;
- fechas con formatos diferentes;
- fechas no válidas.

Vamos a corregirlos paso a paso.

## 4. Detectar valores nulos

`isna()` identifica los valores ausentes.

Si añadimos `.sum()`, podemos contar cuántos valores nulos hay en cada columna.

````python
df.isna().sum()
````

También podemos calcular el porcentaje de valores nulos por columna.

````python
(df.isna().mean() * 100).round(2)
````

## 5. Detectar filas duplicadas

`duplicated()` devuelve `True` para las filas repetidas.

````python
df[df.duplicated()]
````

Eliminamos duplicados con `drop_duplicates()`.

````python
df = df.drop_duplicates().copy()

print("Filas después de eliminar duplicados:", len(df))
````

## 6. Limpiar columnas de texto

Primero eliminamos espacios innecesarios utilizando `str.strip()`.

Después normalizamos algunas columnas para que tengan un formato consistente.

````python
columnas_texto = ["nombre", "ciudad", "email", "metodo_pago"]

for columna in columnas_texto:
    df[columna] = df[columna].astype("string").str.strip()

df.head()
````

Ahora ponemos nombres, ciudades y métodos de pago en un formato más uniforme.

`str.title()` convierte, por ejemplo:

- `BARCELONA` → `Barcelona`
- `tarjeta` → `Tarjeta`

````python
df["nombre"] = df["nombre"].str.title()
df["ciudad"] = df["ciudad"].str.title()
df["metodo_pago"] = df["metodo_pago"].str.title()

df[["nombre", "ciudad", "metodo_pago"]]
````

## 7. Limpiar la columna de edad

La columna `edad` contiene valores como:

- `"25"`
- `"35 años"`
- valores vacíos
- edades imposibles como `200` o `-3`

Primero extraemos la parte numérica y la convertimos a número.

````python
df["edad"] = (
    df["edad"]
    .astype("string")
    .str.extract(r"(\d+)", expand=False)
)

df["edad"] = pd.to_numeric(df["edad"], errors="coerce")

df[["nombre", "edad"]]
````

`errors="coerce"` convierte los valores que no se pueden transformar en número a `NaN`.

Ahora consideramos inválidas las edades menores de 0 o mayores de 100.

````python
df.loc[(df["edad"] < 0) | (df["edad"] > 100), "edad"] = pd.NA

df[["nombre", "edad"]]
````

Podemos rellenar las edades que faltan utilizando, por ejemplo, la **mediana**.

La mediana suele ser más robusta que la media cuando existen valores extremos.

````python
mediana_edad = int(df["edad"].median().round())

df["edad"] = df["edad"].fillna(mediana_edad)

print("Mediana utilizada:", mediana_edad)
df[["nombre", "edad"]]

print("Mediana utilizada:", mediana_edad)
df[["nombre", "edad"]]
````

## 8. Limpiar la columna de importe

El importe contiene varios problemas:

- números con punto decimal;
- números con coma decimal (`89,90`);
- texto (`abc`);
- valores nulos;
- valores negativos.

Primero sustituimos la coma decimal por un punto.

````python
df["importe_compra"] = (
    df["importe_compra"]
    .astype("string")
    .str.replace(",", ".", regex=False)
)

df["importe_compra"] = pd.to_numeric(df["importe_compra"], errors="coerce")

df[["nombre", "importe_compra"]]
````

Los importes negativos no tienen sentido en este ejemplo, por lo que los convertimos en valores nulos.

````python
df.loc[df["importe_compra"] < 0, "importe_compra"] = pd.NA

df[["nombre", "importe_compra"]]
````

Podemos rellenar los importes ausentes con la mediana de la columna.

````python
mediana_importe = df["importe_compra"].median()

df["importe_compra"] = df["importe_compra"].fillna(mediana_importe)

print("Mediana utilizada:", mediana_importe)
````

## 9. Convertir la columna de fecha

Las fechas aparecen en formatos distintos.

`pd.to_datetime()` intenta convertirlas a un tipo de fecha real.

Con `errors="coerce"`, las fechas imposibles se convierten en `NaT` (*Not a Time*).

````python
df["fecha_compra"] = pd.to_datetime(
    df["fecha_compra"],
    errors="coerce",
    dayfirst=True
)

df[["nombre", "fecha_compra"]]
````

Podemos comprobar qué registros tienen una fecha inválida.

````python
df[df["fecha_compra"].isna()]
````

En este ejemplo eliminaremos las filas cuya fecha no se puede recuperar de forma fiable.

````python
df = df.dropna(subset=["fecha_compra"]).copy()
````

## 10. Tratar otros valores ausentes

En este ejemplo:

- si falta la ciudad, escribiremos `"Desconocida"`;
- si falta el email, escribiremos `"No disponible"`.

En un proyecto real, la decisión dependerá del significado de cada variable.

````python
df["ciudad"] = df["ciudad"].fillna("Desconocida")
df["email"] = df["email"].fillna("No disponible")
````

## 11. Comprobar el resultado final

Después de limpiar los datos, conviene revisar de nuevo:

- valores nulos;
- duplicados;
- tipos de datos;
- primeras filas.

````python
print("Valores nulos:")
print(df.isna().sum())

print("\nDuplicados:", df.duplicated().sum())

print("\nDimensiones finales:", df.shape)
````

## 12. Crear alguna variable nueva

Una vez limpio el dataset, podemos empezar a transformarlo.

Por ejemplo, podemos crear una categoría de compra según el importe.

````python
df["categoria_compra"] = pd.cut(
    df["importe_compra"],
    bins=[0, 100, 150, float("inf")],
    labels=["Baja", "Media", "Alta"],
    include_lowest=True
)

df[["nombre", "importe_compra", "categoria_compra"]]
````

## 13. Ejemplos rápidos de análisis

Con el dataset limpio ya podemos hacer análisis sencillos.

````python
# Importe medio por ciudad
df.groupby("ciudad")["importe_compra"].mean().round(2)
````

````python
# Número de compras por método de pago
df["metodo_pago"].value_counts()
````

## 14. Guardar el dataset limpio

Finalmente podemos exportar el DataFrame limpio a un nuevo archivo CSV.

````python
df.to_csv("ventas_clientes_limpio.csv", index=False, encoding="utf-8-sig")

print("Archivo guardado correctamente.")
````

# Resumen

El flujo básico de limpieza de datos con Pandas suele ser:

**Cargar → explorar → detectar errores → limpiar → validar → guardar**

Funciones utilizadas en este ejemplo:

```python
pd.read_csv()
df.head()
df.info()
df.isna()
df.duplicated()
df.drop_duplicates()
str.strip()
str.title()
pd.to_numeric()
pd.to_datetime()
fillna()
dropna()
groupby()
value_counts()
to_csv()
```

Este proceso se conoce habitualmente como **data cleaning** o **data preprocessing**.
